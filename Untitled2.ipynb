```python
import sys
import importlib
importlib.invalidate_caches()

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

try:
    from ucimlrepo import fetch_ucirepo
except ModuleNotFoundError:
    # Fallback to force sys.path refresh if needed
    import subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "ucimlrepo"])
    importlib.invalidate_caches()
    from ucimlrepo import fetch_ucirepo

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN

from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

from sklearn.decomposition import PCA
from sklearn.neighbors import NearestNeighbors

import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
# Fetch the real-world Online Retail dataset
online_retail = fetch_ucirepo(id=352)

df = online_retail.data.features.copy()

print("Dataset loaded successfully!")

print("\nDataset Shape:")
print(df.shape)

print("\nFirst Five Records:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Description:")
display(df.describe())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
# Drop missing values based on available columns
required_columns = [
    'Description',
    'Quantity',
    'InvoiceDate',
    'UnitPrice',
    'CustomerID',
    'Country'
]

df = df.dropna(subset=required_columns).copy()

# Remove invalid quantities and prices
df = df[
    (df['Quantity'] > 0) &
    (df['UnitPrice'] > 0)
].copy()

# Convert date column
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Calculate total sales
df['SalesAmount'] = (
    df['Quantity'] * df['UnitPrice']
)

# Extract date-related features
df['Month'] = df['InvoiceDate'].dt.month

# Rename columns to make them easier to understand
df = df.rename(columns={
    'Description': 'product_name',
    'Quantity': 'quantity',
    'InvoiceDate': 'order_date',
    'UnitPrice': 'unit_price',
    'CustomerID': 'customer_id',
    'Country': 'country',
    'SalesAmount': 'sales_amount',
    'Month': 'month'
})

# Select 10,000 records
df = df.sample(
    n=min(10000, len(df)),
    random_state=42
).reset_index(drop=True)

print("Preprocessing completed!")
print("Processed Dataset Shape:", df.shape)
display(df.head())
```

```python
# Select features
features = df[
    [
        'quantity',
        'unit_price',
        'sales_amount',
        'country',
        'month'
    ]
].copy()

# Numerical and categorical columns
numeric_features = [
    'quantity',
    'unit_price',
    'sales_amount'
]

categorical_features = [
    'country',
    'month'
]

# Preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(
            handle_unknown='ignore',
            sparse_output=False
        ), categorical_features)
    ]
)

# Transform data
X = preprocessor.fit_transform(features)

print("Feature encoding completed!")
print("Processed Feature Shape:", X.shape)

# Select representative sample
X_sample = X[:3000]

print("Clustering Samples:", len(X_sample))
```

```python
# Find optimal K using Silhouette Score
k_values = range(2, 9)
k_scores = []

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    labels = model.fit_predict(X_sample)
    score = silhouette_score(
        X_sample,
        labels,
        sample_size=1000,
        random_state=42
    )
    k_scores.append(score)
    print(f"K = {k}, Silhouette Score = {score:.4f}")

# Select optimal K
optimal_k = list(k_values)[np.argmax(k_scores)]
print("\nOptimal K:", optimal_k)
```

```python
# Train final K-Means model
kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)
kmeans_labels = kmeans.fit_predict(X_sample)

print("K-Means Clustering Completed!")
print("\nCluster Distribution:")
print(pd.Series(kmeans_labels).value_counts().sort_index())
```

```python
# Find optimal number of clusters for Hierarchical Clustering
hierarchical_scores = []

for k in range(2, 9):
    model = AgglomerativeClustering(
        n_clusters=k,
        linkage='ward'
    )
    labels = model.fit_predict(X_sample)
    score = silhouette_score(
        X_sample,
        labels,
        sample_size=1000,
        random_state=42
    )
    hierarchical_scores.append(score)
    print(f"K = {k}, Silhouette Score = {score:.4f}")

optimal_hierarchical_k = list(range(2, 9))[np.argmax(hierarchical_scores)]
print("\nOptimal Clusters:", optimal_hierarchical_k)
```

```python
# Apply Hierarchical Clustering with optimal K
hierarchical = AgglomerativeClustering(
    n_clusters=optimal_hierarchical_k,
    linkage='ward'
)
hierarchical_labels = hierarchical.fit_predict(X_sample)

print("Hierarchical Clustering Completed!")
print("\nCluster Distribution:")
print(pd.Series(hierarchical_labels).value_counts().sort_index())
```

```python
def evaluate_clustering(X, labels, algorithm):
    labels = np.asarray(labels)

    # Remove DBSCAN noise
    if algorithm == "DBSCAN":
        mask = labels != -1
        X_eval = X[mask]
        labels_eval = labels[mask]
    else:
        X_eval = X
        labels_eval = labels

    clusters = len(np.unique(labels_eval))

    if clusters < 2 or clusters >= len(labels_eval):
        return {
            "Algorithm": algorithm,
            "Clusters": clusters,
            "Silhouette Score": np.nan,
            "Davies Bouldin Index": np.nan,
            "Calinski Harabasz Score": np.nan,
            "Noise Points": int(np.sum(labels == -1)) if algorithm == "DBSCAN" else 0
        }

    return {
        "Algorithm": algorithm,
        "Clusters": clusters,
        "Silhouette Score": silhouette_score(X_eval, labels_eval),
        "Davies Bouldin Index": davies_bouldin_score(X_eval, labels_eval),
        "Calinski Harabasz Score": calinski_harabasz_score(X_eval, labels_eval),
        "Noise Points": int(np.sum(labels == -1)) if algorithm == "DBSCAN" else 0
    }
```

```python
# Evaluate all clustering algorithms
results = []
results.append(evaluate_clustering(X_sample, kmeans_labels, "K-Means"))
results.append(evaluate_clustering(X_sample, hierarchical_labels, "Hierarchical"))
try:
    results.append(evaluate_clustering(X_sample, dbscan_labels, "DBSCAN"))
except NameError:
    pass

comparison_df = pd.DataFrame(results)
print("CLUSTERING PERFORMANCE COMPARISON")
display(comparison_df.round(4))
```

```python
# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_sample)

print("PCA completed!")
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)
print("Total Explained Variance:", sum(pca.explained_variance_ratio_))
```

```python
# Plot all clustering results
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

cluster_labels = [
    kmeans_labels,
    hierarchical_labels,
    dbscan_labels if 'dbscan_labels' in globals() else np.zeros(len(X_sample))
]

titles = [
    "K-Means Clustering",
    "Hierarchical Clustering",
    "DBSCAN Clustering"
]

for ax, labels, title in zip(axes, cluster_labels, titles):
    scatter = ax.scatter(
        X_pca[:, 0],
        X_pca[:, 1],
        c=labels,
        cmap='viridis',
        s=12
    )
    ax.set_title(title)
    ax.set_xlabel("Principal Component 1")
    ax.set_ylabel("Principal Component 2")
    plt.colorbar(scatter, ax=ax, label="Cluster")

plt.tight_layout()
plt.show()
```

```python
eps_values = [0.8, 1.0, 1.2, 1.5, 1.8, 2.0]
dbscan_results = []

for eps in eps_values:
    model = DBSCAN(
        eps=eps,
        min_samples=10
    )
    labels = model.fit_predict(X_sample)
    unique_labels = set(labels)
    clusters = len(unique_labels - {-1})
    noise = np.sum(labels == -1)

    # Evaluate only non-noise points
    mask = labels != -1
    if len(set(labels[mask])) >= 2:
        score = silhouette_score(
            X_sample[mask],
            labels[mask],
            sample_size=min(1000, mask.sum()),
            random_state=42
        )
    else:
        score = np.nan

    dbscan_results.append({
        'eps': eps,
        'clusters': clusters,
        'noise': noise,
        'silhouette': score
    })
    print(
        f"EPS={eps}, "
        f"Clusters={clusters}, "
        f"Noise={noise}, "
        f"Silhouette={score:.4f}" if np.isfinite(score) else f"EPS={eps}, Clusters={clusters}, Noise={noise}, Silhouette={score}"
    )
```

```python
# Select the best valid DBSCAN model
valid_results = [
    r for r in dbscan_results
    if np.isfinite(r['silhouette'])
]

if not valid_results:
    raise ValueError(
        "No valid DBSCAN configuration found. "
        "Try a different eps or min_samples."
    )

best_dbscan = max(
    valid_results,
    key=lambda x: x['silhouette']
)
best_eps = best_dbscan['eps']

dbscan = DBSCAN(
    eps=best_eps,
    min_samples=10
)
dbscan_labels = dbscan.fit_predict(X_sample)

print("\nBest DBSCAN Model Selected!")
print("Best EPS:", best_eps)
print("Cluster Distribution:")
print(pd.Series(dbscan_labels).value_counts().sort_index())
```

```python
# Recalculate DBSCAN with safe evaluation
dbscan_labels = dbscan.fit_predict(X_sample)
mask = dbscan_labels != -1

if len(set(dbscan_labels[mask])) >= 2:
    dbscan_silhouette = silhouette_score(
        X_sample[mask],
        dbscan_labels[mask]
    )
    print("DBSCAN Silhouette:", dbscan_silhouette)
else:
    dbscan_silhouette = np.nan
    print("DBSCAN does not have enough valid clusters.")
```
